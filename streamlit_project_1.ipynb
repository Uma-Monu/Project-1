```python
%%writefile pages/2_Analysis.py

import streamlit as st

st.title("Analysis Page")

st.write("SQL Analysis")

import streamlit as st
import sqlite3
import pandas as pd

st.title("Restaurant Analysis")

query = """
SELECT
       cost_for_two,
       ROUND(AVG(rate), 2) AS avg_rating,
       COUNT(*) AS total_restaurants
FROM restaurants
WHERE cost_for_two IS NOT NULL
  AND rate IS NOT NULL
GROUP BY cost_for_two
HAVING COUNT(*) > 5
ORDER BY cost_for_two;
"""

result = pd.read_sql_query(query, conn)
print(result.head(10))
df = pd.read_sql(query, conn)
st.dataframe(df)
conn = sqlite3.connect("uber.db")
```

```python
%%writefile pages/3_Dashboard.py

import streamlit as st
import pandas as pd
import sqlite3
conn = sqlite3.connect("uber.db")
st.title("Restaurant Analysis")

query = """
SELECT
       cost_for_two,
       ROUND(AVG(rate), 2) AS avg_rating,
       COUNT(*) AS total_restaurants
FROM restaurants
WHERE cost_for_two IS NOT NULL
  AND rate IS NOT NULL
GROUP BY cost_for_two
HAVING COUNT(*) > 5
ORDER BY cost_for_two;
"""

result = pd.read_sql_query(query, conn)
st.dataframe(result)
st.bar_chart(
    data=result,
    x="cost_for_two",
    y="avg_rating"
)
conn.close()
```
